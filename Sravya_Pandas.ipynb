## Q1. Array operations and vectorization

```python
import numpy as np
import pandas as pd

# Creating a NumPy array
arr = np.array([10, 20, 30])

# Creating a Pandas Series
s = pd.Series([10, 20, 30])

# NumPy performs element-wise multiplication
print(arr * 2)

# Pandas Series also performs element-wise multiplication
print(s * 2)
```

### 1. Predict the output:

arr * 2 → [20 40 60]
s * 2 → Series containing 20, 40, 60 with index values 0, 1, 2.

### 2. Similarity:
Both NumPy arrays and Pandas Series support vectorized operations, so multiplication by 2 is performed on every element without using an explicit loop.

Difference:
A NumPy array does not normally have labeled indexes, whereas a Pandas Series has an index for each value.

## Q2. Boolean Filtering

```python
import pandas as pd

# Creating a DataFrame containing student names and their marks
df = pd.DataFrame({
    'Student': ['A', 'B', 'C', 'D'],
    'Marks': [45, 80, 65, 90]
})

# Filtering the rows where Marks are greater than or equal to 65
result = df[df['Marks'] >= 65]

# Selecting only the Student and Marks columns
print(result[['Student', 'Marks']])
```

### 1. Predict the rows displayed:

The condition is:

df['Marks'] >= 65

It checks each student's marks:

A → 45 >= 65 → False
B → 80 >= 65 → True
C → 65 >= 65 → True
D → 90 >= 65 → True

Therefore, students B, C, and D are displayed.

### 2. Number of rows and columns:

Rows: 3
Columns: 2

So the resulting DataFrame has 3 rows and 2 columns.

Underlying Pandas operation

df['Marks'] >= 65 creates a Boolean Series:

0    False
1     True
2     True
3     True
Name: Marks, dtype: bool

When this Boolean Series is passed inside df[...], Pandas keeps only the rows where the condition is True. This is called Boolean filtering / Boolean indexing.

## Q4. Group-wise Analysis

```python
import pandas as pd

# Create a DataFrame containing departments and employee salaries
df = pd.DataFrame({
    'Department': ['IT', 'HR', 'IT', 'HR'],
    'Salary': [40000, 30000, 60000, 50000]
})

# Group employees by department and calculate the average salary
# for each department
print(df.groupby('Department')['Salary'].mean())
```

groupby('Department') separates the employees into HR and IT groups. Then ['Salary'].mean() calculates the average salary separately for each department.

HR → (30000 + 50000) / 2 = 40000
IT → (40000 + 60000) / 2 = 50000

### Explanation: groupby() groups rows based on a column and applies an operation such as mean() to each group.

### Q1. Two NumPy arrays have shapes (3, 1) and (1, 3). Predict the shape of their element-wise sum and explain how broadcasting works. Why might the result
differ from adding two one-dimensional arrays?

```python
import numpy as np

# Create a 2D column array with 3 rows and 1 column
a = np.array([[10], [20], [30]])

# Create a 2D row array with 1 row and 3 columns
b = np.array([[1, 2, 3]])

# NumPy broadcasts the two arrays to compatible shapes
# and performs element-wise addition
result = a + b

# Display the resulting array
print(result)

# Display the shape of the resulting array
print(result.shape)
```

### 1. Predict the output
[[11 12 13]
 [21 22 23]
 [31 32 33]]
(3, 3)

### 2. Shapes before addition
a → (3, 1)
b → (1, 3)

### 3 Explain how NumPy broadcasting enables the addition of these arrays.

NumPy broadcasting automatically expands the smaller array to match the shape of the larger array, allowing element-wise addition without manually reshaping the arrays. Here, arrays of shapes (3,1) and (1,3) are broadcast to (3,3), and their corresponding elements are added.

### 4. What would happen if b were replaced with np.array([1, 2])? Explain the reason.

a → (3, 1)
b → (2,)

NumPy aligns dimensions from the right side:

a → (3, 1)
b →    (2)

The dimensions 1 and 2 are not compatible because neither is equal nor is either dimension 1 in the aligned comparison. Therefore, NumPy raises a ValueError instead of producing a result.

ValueError: operands could not be broadcast together with shapes (3,1) (2,)

### Explanation: 
(3,1) and (2,) cannot be broadcast together because their corresponding dimensions are incompatible.

### Q2. A CSV file contains commas inside quoted text fields, missing values, and dates in different formats. Explain how incorrect parsing can affect the
resulting DataFrame and identify suitable read options or preprocessing steps.



```python
import numpy as np

a = np.array([[10], [20], [30]])
b = np.array([[1, 2, 3]])

# Broadcasting expands the compatible dimensions during element-wise addition.
result = a + b

print(result)
print("Shape:", result.shape)

# These shapes are incompatible for broadcasting.
try:
    a + np.array([1, 2])
except ValueError as error:
    print("Error:", error)
```

### 1. How is "Asha, R" handled?

The comma inside "Asha, R" is correctly treated as part of the Name field because the value is enclosed in double quotes.

So Pandas reads it as:

ID   Name      Date        Marks
101  Asha, R   2026-09-01  80

It does not create an extra column for the comma.

### 2. Potential issues with Date and Marks columns
Date: The dates are in different formats: 2026-09-01 and 01/09/2026. They should be converted to a consistent datetime format.
Marks: The column contains missing values: one blank value and one "NA". These should be handled appropriately before numerical analysis.

### 3. Corrected code

```python
import pandas as pd
from io import StringIO

csv_data = '''ID,Name,Date,Marks
101,"Asha, R",2026-09-01,80
102,Ravi,01/09/2026,
103,Neha,2026-09-03,NA'''

# Quoted commas stay inside the Name field.
# Blank values and NA are treated as missing values for inspection.
df = pd.read_csv(
    StringIO(csv_data),
    na_values=['NA', ''],
    keep_default_na=True
)

# Convert the mixed date representations into datetime values.
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True, errors='coerce')

print(df)
print("\nData types:")
print(df.dtypes)
```

### 1. Load the CSV file and inspect the dataset using head(), info(), and isnull().sum().

```python
import pandas as pd

# Load the student assessment dataset from the CSV file
df = pd.read_csv("student_assessment_dirty.csv")

# Display the first five rows of the dataset
print("First 5 rows:")
print(df.head())

# Display information about columns, data types, and missing values
print("\nDataset Information:")
df.info()

# Count the number of missing values in each column
print("\nMissing Values:")
print(df.isnull().sum())
```

### Explanation
head() displays the first five records.
info() shows column names, data types, and non-null values.
isnull().sum() counts missing values in each column.

### 2. Standardize department names so that IT, it, and I.T. are represented consistently. Apply the same approach to Finance.

```python
# Remove extra spaces and convert all department names to uppercase
df["Department"] = df["Department"].str.strip().str.upper()

# Replace different forms of department names with standard names
df["Department"] = df["Department"].replace({
    "IT": "IT",
    "I.T.": "IT",
    "HR": "HR",
    "FINANCE": "Finance"
})

# Display the standardized department names
print(df["Department"].unique())
```

### Explanation

str.upper() makes different letter cases consistent, and replace() converts variations such as it, I.T., and finance into standard department names.

### 3. Identify and remove exact duplicate student assessment record

```python
# Count the number of duplicate rows in the dataset
duplicate_count = df.duplicated().sum()

# Display the number of duplicate records
print("Number of duplicate records:", duplicate_count)

# Remove exact duplicate rows
df = df.drop_duplicates()

# Display the number of rows after removing duplicates
print("Number of records after removing duplicates:", len(df))
```

### Explanation

duplicated() identifies rows that are completely identical, while drop_duplicates() removes the duplicate record.

### 4. Convert the Marks column into a numeric format, handling "88 marks", "Absent", "NA", and blank entries appropriately.

```python
# Keep a copy of the original Marks column
# so that values such as "Absent" can be investigated later
df["Original_Marks"] = df["Marks"]

# Remove the word "marks" from values such as "88 marks"
df["Marks"] = (
    df["Marks"]
    .astype("string")
    .str.replace("marks", "", case=False, regex=False)
    .str.strip()
)

# Convert the cleaned Marks column into numeric values
# Non-numeric values such as "Absent" become NaN
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

# Display the Marks column after conversion
print(df[["Original_Marks", "Marks"]])
```

### Explanation
"88 marks" becomes 88.
"Absent" becomes NaN.
"NA" is treated as missing.
Blank values remain missing.
Numeric values such as 85, 78, and 92 remain numbers.

### 5. Investigate missing marks and distinguish absent students from students whose marks have not been entered. State one risk of deleting missing values
without investigating why they are missing.

```python
# Identify students who were explicitly marked as absent
absent_students = df[
    df["Original_Marks"].astype("string").str.strip().str.lower() == "absent"
]

# Display absent students
print("Absent Students:")
print(absent_students[["Student_ID", "Student_Name", "Original_Marks"]])

# Identify records where the cleaned Marks value is missing
missing_marks = df[df["Marks"].isna()]

# Display all records with missing marks
print("\nStudents with Missing Marks:")
print(missing_marks[["Student_ID", "Student_Name", "Original_Marks"]])
```

### Explanation

1.Absent should be retained as a separate status instead of treating it exactly like a mark that was never entered.

2.Risk of deleting missing values

3.Deleting missing values without investigating why they are missing may remove useful information and can lead to biased or incorrect analysis.

### 6. Identify marks outside the valid range of 0–100 and decide how they should be handled

```python
# Find marks below 0 or above 100
invalid_marks = df[
    (df["Marks"] < 0) | (df["Marks"] > 100)
]

# Display the invalid records
print("Invalid Marks:")
print(invalid_marks[["Student_ID", "Student_Name", "Marks"]])
```

### Handle the invalid mark

```python
# Replace marks outside the valid range with a missing value
# until the original record is verified
df.loc[
    (df["Marks"] < 0) | (df["Marks"] > 100),
    "Marks"
] = pd.NA
```

### Explanation

105 should not automatically be deleted. It should first be investigated because it could be a data-entry error or a special case. Until it is verified, treating it as missing prevents it from affecting calculations.

### 7. Display the cleaned DataFrame and report the number of records removed, missing marks remaining, and invalid marks identified.

```python
# Display the cleaned DataFrame
print("Cleaned DataFrame:")
print(df)

# Count the number of records removed
print("\nRecords removed:", duplicate_count)

# Count the number of missing marks remaining
print("Missing marks remaining:", df["Marks"].isna().sum())

# Count the number of invalid marks identified
print("Invalid marks identified:", len(invalid_marks))
```

### Explanation

The dataset is cleaned by standardizing department names, removing exact duplicates, converting marks into numeric values, investigating missing/absent records, and identifying invalid marks outside the 0–100 range.
