## Q1. NumPy Broadcasting

```python
import numpy as np

a = np.array([[10], [20], [30]])
b = np.array([[1, 2, 3]])

# Broadcasting expands the compatible dimensions during element-wise addition.
result = a + b

print(result)
print("Shape:", result.shape)

# These shapes are incompatible for broadcasting.
try:
    a + np.array([1, 2])
except ValueError as error:
    print("Error:", error)
```

**Explanation:** Shapes `(3, 1)` and `(1, 3)` broadcast to `(3, 3)`. A `(2,)` array cannot be aligned with the required dimensions, so NumPy raises a broadcasting error.

## Q2. CSV Parsing, Missing Values, and Dates

```python
import pandas as pd
from io import StringIO

csv_data = '''ID,Name,Date,Marks
101,"Asha, R",2026-09-01,80
102,Ravi,01/09/2026,
103,Neha,2026-09-03,NA'''

# Quoted commas stay inside the Name field.
# Blank values and NA are treated as missing values for inspection.
df = pd.read_csv(
    StringIO(csv_data),
    na_values=['NA', ''],
    keep_default_na=True
)

# Convert the mixed date representations into datetime values.
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True, errors='coerce')

print(df)
print("\nData types:")
print(df.dtypes)
```

**Explanation:** The quoted comma in `Asha, R` is handled correctly by CSV parsing. The blank and `NA` Marks values need investigation, and the Date column is converted to a datetime type for consistent analysis.

## Q3. Extreme Salary Value

```python
import pandas as pd

df = pd.DataFrame({
    "Employee": ["A", "B", "C", "D"],
    "Salary": [25000, 30000, 35000, 10000000]
})

print("Mean:", df["Salary"].mean())
print("Median:", df["Salary"].median())

# Locate the record with the highest salary for investigation.
highest_salary_record = df.loc[df["Salary"].idxmax()]

print("\nHighest salary record:")
print(highest_salary_record)
```

**Explanation:** The extreme salary increases the mean substantially, while the median remains resistant to the extreme value. The median better represents a typical salary here. The highest-salary record should be investigated before removal because it may be genuine.

## Q1. Student Assessment Data Cleaning

```python
import pandas as pd
import numpy as np
from io import StringIO

csv_data = '''Student_ID,Student_Name,Department,Assessment,Marks
S101,Asha,IT,Python,85
S102,Ravi,it,Python,78
S103,Neha,I.T.,Python,92
S104,Karan,HR,Python,
S105,Meera,HR,Python,Absent
S106,Arjun,Finance,Python,0
S107,Priya,finance,Python,67
S108,Rahul,IT,Python,75
S102,Ravi,it,Python,78
S109,Sneha,I.T.,Python,NA
S110,Varun,HR,Python,88 marks
S111,Diya,Finance,Python,91
S112,Aman,IT,Python,
S113,Nisha,HR,Python,64
S114,Kabir,Finance,Python,105'''

# Load the dirty dataset.
df = pd.read_csv(StringIO(csv_data), keep_default_na=False)

print("First five rows:")
print(df.head())

print("\nData information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())
```

**Explanation:** We first inspect the dataset before changing it. `head()`, `info()`, and `isnull().sum()` reveal its structure, data types, and missing values.

```python
# Standardize department names so equivalent labels become consistent.
df['Department'] = (
    df['Department']
    .str.strip()
    .str.lower()
    .replace({
        'it': 'IT',
        'i.t.': 'IT',
        'finance': 'Finance',
        'hr': 'HR'
    })
)

# Remove exact duplicate assessment records.
before = len(df)
df = df.drop_duplicates()
duplicates_removed = before - len(df)

print("Duplicates removed:", duplicates_removed)
print("\nStandardized departments:")
print(sorted(df['Department'].unique()))
```

**Explanation:** Standardizing labels prevents the same department from being treated as multiple groups. Exact duplicate rows are removed and the number removed is recorded.

```python
# Keep track of whether the original value represented an absence.
df['Attendance_Status'] = np.where(
    df['Marks'].astype(str).str.strip().str.lower().eq('absent'),
    'Absent',
    'Recorded'
)

# Extract numeric marks such as 88 from "88 marks".
# Values such as blank, NA, and Absent become NaN.
df['Marks'] = (
    df['Marks']
    .astype(str)
    .str.extract(r'(\d+(?:\.\d+)?)', expand=False)
    .astype(float)
)

print(df[['Student_ID', 'Student_Name', 'Marks', 'Attendance_Status']])
```

**Explanation:** Numeric marks are extracted from text values, while `Absent`, `NA`, and blank values remain unavailable for numerical analysis. Absence is kept as a separate status instead of silently converting it to zero.

```python
# Investigate missing marks separately.
missing_records = df[df['Marks'].isna()]

print("Missing/unavailable marks:")
print(missing_records[['Student_ID', 'Student_Name', 'Marks', 'Attendance_Status']])

# Identify marks outside the valid assessment range.
invalid_records = df[(df['Marks'] < 0) | (df['Marks'] > 100)]

print("\nInvalid marks outside 0-100:")
print(invalid_records[['Student_ID', 'Student_Name', 'Marks']])
```

**Explanation:** Missing values are investigated instead of being deleted immediately. Marks outside 0–100 are flagged as invalid and require review before any removal or correction.

```python
# Label each record so the cleaning result remains transparent.
df['Mark_Status'] = np.select(
    [
        df['Marks'].isna(),
        (df['Marks'] < 0) | (df['Marks'] > 100)
    ],
    [
        'Missing/Unavailable',
        'Invalid'
    ],
    default='Valid'
)

print("Cleaned DataFrame:")
print(df)

print("\nFinal summary:")
print("Records removed:", duplicates_removed)
print("Missing marks remaining:", int(df['Marks'].isna().sum()))
print("Invalid marks identified:", int(((df['Marks'] < 0) | (df['Marks'] > 100)).sum()))
```

**Explanation:** The final result keeps the cleaning decisions visible. This makes the dataset easier to audit and prevents meaningful cases such as absence from being confused with an ordinary missing entry.
